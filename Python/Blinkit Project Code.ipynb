```python
sheet_table_map = {
    "customers": "customers",
    "customer_feedback": "customer_feedback",
    "delivery": "delivery",
    "inventory": "inventory",
    "marketing": "marketing",
    "orders": "orders",
    "order_items": "order_items",
    "products": "products"
}
```

```python
from sqlalchemy import create_engine
import pandas as pd

username = "root"
password = "root"
host = "localhost"
port = "3306"
database = "blinkit_project"

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)
```

```python
excel_path = r"D:\blinkit_project\docs\blinkit.xlsx"   # change if needed

for sheet_name, table_name in sheet_table_map.items():
    df = pd.read_excel(excel_path, sheet_name=sheet_name)

    df.to_sql(
        table_name,
        engine,
        if_exists="replace",   # use "append" if you want incremental loads
        index=False
    )

    print(f"Inserted {len(df)} rows into `{table_name}`")
```
